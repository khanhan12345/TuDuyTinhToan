w4a26 nhap so nguyen duong a, tim so trong day fibonacci lon nhat nhung khong vuot qua a

```python
a=int(input())
n=0
def fibo(n):
    if n==0:
        return 0
    if n==1:
        return 1
    if n==2:
        return 1
    if n>2:
        return fibo(n-1) + fibo(n-2)
    
while fibo(n)<=a:
    n=n+1
print(fibo(n-1))
```

w4a29 cho mot chuoi chi gom ky tu chu cai thuong va chu cai hoa, doi ky tu hoa sang thuong va ky tu thuong sang hoa

```python
s=input()
chuoi=[]
for ch in s:
    if ch.islower():
        chuoi.append(ch.upper())
    if ch.isupper():
        chuoi.append(ch.lower())
print(*chuoi, sep='')
```

w4a31 nhap vao mot chuoi, dem co bao nhieu in thuong, in hoa, so

```python
s=input()
inthg=0
inhoa=0
so=0
for ch in s:
    if ch.isalpha():
        if ch.isupper():
            inhoa=inhoa+1
        if ch.islower():
            inthg=inthg+1
    if ch.isdigit():
        so=so+1
print(inthg, inhoa, so)
```
