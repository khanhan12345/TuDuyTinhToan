w10a1 divide by zero

```python
a=int(input())
b=int(input())

def divide(a,b):
    return f'{a/b:.2f}'

try:
    print(divide(a,b))
except ZeroDivisionError:
    print('loi chia cho 0')
```

w10a4 always print finally

```python
try:
    a=float(input('nhap so thu nhat: '))
    b=float(input('nhap so thu hai: '))
except ValueError:
    print('nhap cho dung de bai vao')
else:
    print(a+b)
finally:
    print('okok')
```

w10a10 nhap so nguyen n va mot mang n so nguyen duong, mot mang hop le neu cac chu so trong do khac nhau, in ra tong phan tu neu hop le, khong thi in khong hop le

```python
try:
    n=int(input('nhap so luong: '))
    if n<=0:
        print('nhap so nguyen duong')
    else:
        arr=list(map(int, input().split()))
        if len(arr)!=n:
            print('nhap dung n phan tu')
        else:
            soam=False
            for ch in arr:
                if ch<0:
                    soam=True
                    break
            if soam:
                print('nhap phan tu nguyen duong vao')
            else:
                ok=set(arr)
                if len(ok)==len(arr):
                    tong=sum(arr)
                    print(f'tong la {tong}')
                else:
                    print('mang co phan tu trung lap')
except ValueError:
    print('nhap dung kieu so nguyen')
    
            
        
```

w10a13 pop element

```python
arr=list(map(int, input().split()))
k=int(input())
try:
    
    if k<0 or k>len(arr)-1:
        print('chi so khong hop le')
    else:
        print(f'phan tu cbi xoa la {arr[k]}')
        arr.pop(k)
except IndexError:
    print('chi so khong hop le hahahahah')
        
```

```python
sum=0
arr=list(map(str, input().split()))
for ch in arr:
    try:
        chh=int(ch)
        sum=sum+chh
    except ValueError:
        continue 
print(sum)
```
