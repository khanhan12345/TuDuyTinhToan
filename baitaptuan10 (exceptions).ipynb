w10a1 divide by zero

```python
a=int(input())
b=int(input())

def divide(a,b):
    return f'{a/b:.2f}'

try:
    print(divide(a,b))
except ZeroDivisionError:
    print('loi chia cho 0')
```

```python
n=int(input('nhap so phan tu'))
lst=list(map(int, input().split()))
try:
    print(lst[n])
except IndexError:
    print('truy cap sai r')
```

w10a4 always print finally

```python
try:
    a=float(input('nhap so thu nhat: '))
    b=float(input('nhap so thu hai: '))
except ValueError:
    print('nhap cho dung de bai vao')
else:
    print(a+b)
finally:
    print('okok')
```

w10a10 nhap so nguyen n va mot mang n so nguyen duong, mot mang hop le neu cac chu so trong do khac nhau, in ra tong phan tu neu hop le, khong thi in khong hop le

```python
n=int(input())
try:
    lst=[]
    for _ in range(n):
        k=int(input())
        lst.append(k)

    print(sum(lst))
except TypeError:
    print('nhap dung so nguyen vao')
except ValueError:
    print('nhap cho dung vao')
```
