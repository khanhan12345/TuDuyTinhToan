bmi check

```python
def bmi_compare(w1, h1, w2, h2):
    if (w1/(h1*h1)) > (w2/(h2*h2)):
        return 1
    if (w2/(h2*h2)) > (w1/(h1*h1)):
        return 2
    if (w2/(h2*h2)) == (w1/(h1*h1)):
        return 0
    


assert bmi_compare(70, 1.75, 70, 1.75)==0
print('ok')
```

math grade

```python
def grade(s):
    if s>=9:
        return 'A+'
    elif s>=8.5:
        return 'A'
    elif s>=8:
        return 'B+'
    elif s>=7:
        return 'B'
    else:
        return 'E'
    
assert(grade(8.3))=='B+'
print('ok')
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation

# Cấu hình các thông số ban đầu
r = 1.0  # Bán kính đường tròn
theta_max = 2 * np.pi  # Quay hết 1 vòng dây
frames = 120  # Số khung hình mô phỏng

fig, ax = plt.subplots(figsize=(7, 7))
ax.set_xlim(-1.5 * r, 3 * r)
ax.set_ylim(-2 * r, 2.5 * r)
ax.set_aspect('equal')
ax.grid(True, linestyle='--', alpha=0.6)
ax.set_title("Mô phỏng đường khai triển của đường tròn (Involute)", fontsize=12)

# 1. Vẽ đường tròn cố định (màu đen nét đứt)
t_circle = np.linspace(0, 2 * np.pi, 200)
ax.plot(r * np.cos(t_circle), r * np.sin(t_circle), 'k--', label="Đường tròn bán kính r")

# 2. Khởi tạo các đối tượng đồ họa thay đổi theo thời gian
line_string, = ax.plot([], [], 'r-', linewidth=2, label="Sợi dây căng (TP)")  # Sợi dây màu đỏ
point_P, = ax.plot([], [], 'ro', markersize=6)  # Đầu dây P
point_T, = ax.plot([], [], 'ko', markersize=4)  # Điểm tiếp xúc T
path_P, = ax.plot([], [], 'b-', linewidth=2, label="Quỹ tích điểm P")  # Đường involute màu xanh

# Lưu lại lịch sử tọa độ của P để vẽ đường cong tích lũy
x_history, y_history = [], []

ax.legend(loc="upper left")

def update(frame):
    # Tính góc theta hiện tại tương ứng với khung hình
    theta = (frame / frames) * theta_max
    
    # Tọa độ điểm tiếp xúc T
    x_T = r * np.cos(theta)
    y_T = r * np.sin(theta)
    
    # Tọa độ đầu dây P dựa theo phương trình tham số chứng minh được
    x_P = r * (np.cos(theta) + theta * np.sin(theta))
    y_P = r * (np.sin(theta) - theta * np.cos(theta))
    
    # Cập nhật lịch sử quỹ tích đường đi của P
    x_history.append(x_P)
    y_history.append(y_P)
    
    # Cập nhật dữ liệu hiển thị cho các đường thẳng/điểm trên đồ thị
    line_string.set_data([x_T, x_P], [y_T, y_P])
    point_T.set_data([x_T], [y_T])
    point_P.set_data([x_P], [y_P])
    path_P.set_data(x_history, y_history)
    
    return line_string, point_T, point_P, path_P

# Tạo hiệu ứng chuyển động mượt mà
ani = animation.FuncAnimation(fig, update, frames=frames, interval=40, blit=True, repeat=True)

plt.show()
```
