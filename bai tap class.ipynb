```python
class Cylinder:
    def __init__(self, rr, hh):
        self.r=rr
        self.h=hh
    def getarea(self) -> float:
        return (2*3.14*self.r*self.h) + 2*(self.r)**2*3.14
    def getvolume(self) -> float:
        return 3.14*(self.r)**2*self.h
    
m,n=map(float, input().split())
cy1=Cylinder(m,n)
print(cy1.getarea())
```

```python
class MyDate:
    def __init__(self, dd, mm, yy):
        self.d=dd
        self.m=mm
        self.y=yy
    def check_valid_date(self) -> bool:
        maxday=-1
        match self.m:
            case 1|3|5|7|8|10|12:
                maxday=31
            case 4|6|9|11:
                maxday=30
            case 2:
                if ((self.y) %4==0 and self.y %100!=0) or self.y%400==0:
                    maxday=29
                else:
                    maxday=28
        if self.y>0 and 1<=self.d<=maxday:
            return True
        else:
            return False
        #return self.y>0 and 1<=self.day<=maxday
    def __str__(self):
        return f'{self.d:02d}/{self.m:02d}/{self.y}'
    def next_day(self) ->MyDate:
        nextDay=MyDate(self.d+1, self.m, self.y)
        if nextDay.check_valid_date()==True:
            return nextDay
        nextDay.d=1
        nextDay.m+=1
        if nextDay.check_valid_date():
            return nextDay
        nextDay.m=1
        nextDay.y+=1
        return nextDay




d,m,y=map(int, input().split('/'))
datee=MyDate(d,m,y)


print(datee.check_valid_date())
print(datee.next_day())
```

w8a3 point

```python
class Point:
    def __init__(self, x, y):
        self.x=x
        self.y=y
    def check(self):
        if self.x==0 and self.y==0:
            return 'goc toa do'
        if self.x==0 and self.y!=0:
            return 'nam tren truc tung'
        if self.x!=0 and self.y==0:
            return 'nam tren truc hoanh'
        else:
            return 'khong nam tren truc toa do'
    def distance(self):
        return ((self.x)**2+(self.y)**2)**0.5
    def __str__(self):
        return f'{self.check()}, {self.distance()}'




x,y=map(float, input().split())
abc=Point(x,y)
print(abc)
```

w8a4 caculator

```python
class Caculator:
    def __init__(self, a, b):
        self.a=a
        self.b=b
    def cong(self):
        return self.a + self.b
    def tru(self):
        return self.a - self.b
    def nhan(self):
        return self.a * self.b
    def chia(self):
        return self.a / self.b
    def mu(self):
        return self.a ** self.b
    def du(self):
        return self.a % self.b
    def set_numbers(self, c, d):
        self.a=c
        self.b=d
    
num1=float(input())
num2=float(input())
cacu=Caculator(num1, num2)


while True:
    choice=input()
    if choice=='cong':
        print(cacu.cong())
    if choice=='tru':
        print(cacu.tru())
    if choice=='nhan':
        print(cacu.nhan())

    if choice=='set_numbers':
        newa=float(input('nhap a moi: '))
        newb=float(input('nhap b moi: '))
        cacu.set_numbers(newa, newb)
        print('reset lai so')
    quit_choice=input()
    if quit_choice.lower()=='yes':
        break
```

```python
class ShoppingCart:
    def __init__(self, ):
        self.lst=[]
    def addproduct(self, name, price):
        self.lst.append({'name':name,'price':price})
    def removeproduct(self,name):
        found=False
        for item in self.lst:
            if item['name'].lower()==name.lower():
                self.lst.remove(item)
                found=True
                break
        if not found:
            print('ko thay de xoa')
    def check_empty(self):
        return len(self.lst)==0
    def totalprice(self):
        total=0
        for item in self.lst:
            total=total+item['price']
        return total
    def display_cart(self):
        if self.check_empty():
            print('gio hang rong')
            return
        for idx,item in enumerate(self.lst, start=1):
            print(f'{idx}, ten: {item['name']}, gia: {item['price']}')
    def clear_cart(self):
        self.lst.clear()
        print('wipe out')



```
