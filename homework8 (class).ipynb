w8a16. dung lop shape voi phuong thuc area()
dung lop rectangle (w,h) va circle(r) ke thua tu shape, ghi de area(), lay pi =3.14

```python
import math
class Shape:
    def __init__(self):
        pass
    def area(self) -> float:
        return 1
class Rectangle(Shape):
    def __init__(self, w, h):
        super().__init__()
        self.w=w
        self.h=h
    def area(self) -> float:
        return self.w*self.h 
class Circle(Shape):
    def __init__(self, r):
        super().__init__()
        self.r=r
    def area(self) -> float:
        return 3.14*self.r
    

loaihinh=input()
thamso=list(map(float, input().split()))
hinh=None
if loaihinh=="RECTANGLE":
    w=thamso[0]
    h=thamso[1]
    hinh=Rectangle(w,h)
elif loaihinh== "Circle":
    r=thamso[0]
    hinh=Circle(r)

s=hinh.area()
print(s)

```

w8a17 racing car

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand=brand
        self.speed=speed
    def info(self):      # , brand, speed
        return f'{self.brand} and {self.speed}'

class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        super().__init__(brand, speed)
        self.seats=seats
    def info(self):     # , brand, speed, seats
        return f'{self.brand}, {self.speed} and {self.seats}'
    
mycar=None
dong1=input()
dong2=input().split()


if dong1=='Vehicle':
    brand=dong2[0]
    speed=dong2[1]
    mycar=Vehicle(brand,speed)
if dong1=="Car":
    brand=dong2[0]
    speed=dong2[1]
    seats=dong2[2]
    mycar=Car(brand,speed,seats)

if mycar is not None:
    print(mycar.info())




```

```python
class Person:
    def __init__(self, name, age):
        self.name=name
        self.age=age
    def describe(self):
        return f'{self.name} and {self.age}'
class Student(Person):
    def __init__(self, name, age, school):
        super().__init__()
        self.school=school
    def describe(self):
        return f'{self.name}, {self.age} and {self.school}'
guy=None
dong1=input()
dong2=input().split()
if dong1=="Person":
    name=dong2[0]
    age=dong2[1]
    guy=Person(name, age)
    print(guy.describe())
if dong1=="Student":
    name=dong2[0]
    age=dong2[1]
    school=dong2[2]
    guy=Student(name, age, school)
    print(guy.describe())

```
