w8a16. dung lop shape voi phuong thuc area()
dung lop rectangle (w,h) va circle(r) ke thua tu shape, ghi de area(), lay pi =3.14

```python
import math
class Shape:
    def __init__(self):
        pass
    def area(self) -> float:
        return 1
class Rectangle(Shape):
    def __init__(self, w, h):
        super().__init__()
        self.w=w
        self.h=h
    def area(self) -> float:
        return self.w*self.h 
class Circle(Shape):
    def __init__(self, r):
        super().__init__()
        self.r=r
    def area(self) -> float:
        return 3.14*self.r
    

loaihinh=input()
thamso=list(map(float, input().split()))
hinh=None
if loaihinh=="RECTANGLE":
    w=thamso[0]
    h=thamso[1]
    hinh=Rectangle(w,h)
elif loaihinh== "Circle":
    r=thamso[0]
    hinh=Circle(r)

s=hinh.area()
print(s)

```
